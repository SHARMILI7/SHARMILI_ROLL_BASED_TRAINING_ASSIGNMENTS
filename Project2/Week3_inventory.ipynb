```python
from pyspark.sql import SparkSession
from pyspark.sql.functions import col, when

spark=SparkSession.builder.appName("Week3_Warehouse_Insights").getOrCreate()

df=spark.read.csv("cleaned_stock.csv",header=True,inferSchema=True)

print("===== CLEANED STOCK INPUT =====")
df.show()

df=df.withColumn("stock_status",when(col("current_stock")<col("reorder_level"),"Understocked").when(col("current_stock")>col("reorder_level")*3,"Overstocked").otherwise("Balanced"))

print("===== WAREHOUSE STOCK STATUS =====")
df.show()

print("===== UNDERSTOCKED PRODUCTS =====")
df.filter(col("stock_status")=="Understocked").show()

print("===== OVERSTOCKED PRODUCTS =====")
df.filter(col("stock_status")=="Overstocked").show()

df.toPandas().to_csv("warehouse_stock_status.csv",index=False)

print("warehouse_stock_status.csv exported successfully")
```
